```python
import numpy as np

eeg = np.load("cleaned_eeg.npy")
labels = np.load("cleaned_labels.npy")

print("EEG shape:", eeg.shape)
print("Labels shape:", labels.shape)
```

```python
eeg_ml = eeg.reshape(-1, 32, 8064)

print(eeg_ml.shape)
```

```python
labels_ml = labels.reshape(-1, 4)

print(labels_ml.shape)
```

```python
print(eeg_ml[0].shape)
print(labels_ml[0])
```

```python
valence = labels_ml[:, 0]
y = (valence > 5).astype(int)

print(y.shape)
print(np.unique(y))
```

```python
np.save("eeg_ml.npy", eeg_ml)
np.save("labels_ml.npy", labels_ml)
np.save("valence_binary.npy", y)
```
