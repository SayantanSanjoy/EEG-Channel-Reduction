```python
import glob
import pickle
import numpy as np

all_labels = []

for file in sorted(glob.glob("s*.dat")):
    with open(file, "rb") as f:
        d = pickle.load(f, encoding="latin1")
        all_labels.append(d["labels"])

all_labels = np.concatenate(all_labels)

print("Shape:", all_labels.shape)
print("Minimum:", all_labels.min(axis=0))
print("Maximum:", all_labels.max(axis=0))
print("Mean:", all_labels.mean(axis=0))
```

```python
valence = all_labels[:, 0]

y = (valence > 5).astype(int)

print(np.bincount(y))
```

```python
import glob
import pickle
import numpy as np

for file in sorted(glob.glob("s*.dat")):
    with open(file, "rb") as f:
        d = pickle.load(f, encoding="latin1")

    eeg = d["data"][:, :32, :]

    print(
        file,
        eeg.shape,
        np.isnan(eeg).sum(),
        np.isinf(eeg).sum()
    )
```

```python
label_names = ["Valence", "Arousal", "Dominance", "Liking"]

for i, name in enumerate(label_names):

    print("="*40)
    print(name)

    print("Minimum :", all_labels[:, i].min())
    print("Maximum :", all_labels[:, i].max())
    print("Mean    :", all_labels[:, i].mean())
```

```python
# Count how many values are below 1 in each label

for i, name in enumerate(label_names):

    count = np.sum(all_labels[:, i] < 1)

    print(f"{name}: {count} values below 1")
```

```python
# Extract Valence labels

valence = all_labels[:, 0]

# Convert into binary classes
# High Valence (>5) -> 1
# Low Valence (<=5) -> 0

y = (valence > 5).astype(int)

print("Class Distribution:")
print(np.bincount(y))

# If the dataset is balanced, both classes should be reasonably close.
```
