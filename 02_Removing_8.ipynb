```python
import pickle
import numpy as np
import os
```

```python
print(os.getcwd())
```

```python
all_eeg = []

all_labels = []
```

```python
for i in range(1, 33):

    filename = f"s{i:02d}.dat"

    print("Loading:", filename)

    with open(filename, "rb") as file:

        subject = pickle.load(file, encoding="latin1")

    # Keep only EEG channels
    eeg_data = subject["data"][:, :32, :]

    # Remove DC offset
    eeg_data = eeg_data - np.mean(
        eeg_data,
        axis=2,
        keepdims=True
    )

    # Normalize
    mean = np.mean(eeg_data)

    std = np.std(eeg_data)

    eeg_data = (eeg_data - mean) / std

    # Store cleaned EEG
    all_eeg.append(eeg_data)

    # Store labels
    all_labels.append(subject["labels"])
```

```python
all_eeg = np.array(all_eeg)

all_labels = np.array(all_labels)
valence = all_labels[:, :, 0]

print("Mean:", np.mean(valence))
print("Min :", np.min(valence))
print("Max :", np.max(valence))

print("<=5 :", np.sum(valence <= 5))
print(">5  :", np.sum(valence > 5))
```
