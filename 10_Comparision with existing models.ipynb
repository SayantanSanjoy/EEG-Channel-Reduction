# Comparative Analysis of EEG Emotion Recognition Models
## Channel Reduction, Model Comparison, and Comparison with Existing Research

This notebook presents a comprehensive analysis of three proposed EEG emotion recognition pipelines:

1. DGCNN
2. Random Forest with Mean-Variance features
3. Random Forest with Band-Power features

The three proposed pipelines are evaluated under a common experimental framework using:

- The same DEAP-derived EEG dataset
- The same four emotional dimensions:
  - Valence
  - Arousal
  - Dominance
  - Liking
- The same channel configurations:
  - 32 channels
  - 16 channels
  - 8 channels
  - 4 channels

The Random Forest models were retrained to align with the experimental setup used for the DGCNN experiments.

The primary objective is to investigate the relationship between:

    EEG Channel Count
            ↓
    Feature Representation
            ↓
    Model Architecture
            ↓
    Emotion Recognition Performance

The analysis also compares the proposed framework with three existing research approaches:

- HA-CapsNet
- WST-LDA-KNN
- EmoCaps

The main focus is not only on classification accuracy, but also on:

- Explicit EEG channel reduction
- Channel efficiency
- Model architecture comparison
- Feature representation comparison
- Emotion-specific channel requirements
- Potential practical implications for reduced-electrode EEG systems

# Research Questions

This analysis aims to answer the following research questions:

1. Which of the three proposed pipelines performs best for each emotional dimension?

2. Does increasing the number of EEG channels always improve classification performance?

3. Can reduced-channel configurations achieve performance comparable to the full 32-channel configuration?

4. Does the optimal number of EEG channels depend on the emotional dimension?

5. Does the optimal channel configuration depend on the model architecture or feature representation?

6. How does the DGCNN compare with simpler Random Forest pipelines?

7. What is the effect of using statistical Mean-Variance features compared with frequency-domain Band-Power features?

8. How does the proposed framework compare with existing approaches reported in the literature?

9. What advantages does the proposed framework provide beyond absolute classification accuracy?

# Experimental Framework

The proposed study evaluates three different EEG emotion recognition pipelines:

### Model 1: DGCNN

A graph-based deep learning architecture designed to model relationships between EEG channels.

### Model 2: Random Forest + Mean-Variance

A classical machine learning pipeline using statistical features derived from the EEG signals.

### Model 3: Random Forest + Band Power

A classical machine learning pipeline using frequency-domain EEG features derived from different frequency bands.

All three pipelines are evaluated using four channel configurations:

| Configuration | Number of Channels | Channel Reduction |
|---------------|-------------------|-------------------|
| Full | 32 | 0% |
| Reduced | 16 | 50% |
| Reduced | 8 | 75% |
| Reduced | 4 | 87.5% |

The same four emotional dimensions are considered:

- Valence
- Arousal
- Dominance
- Liking

The objective is to determine whether significant channel reduction can be achieved while retaining useful emotion recognition performance.

# Channel Reduction Analysis

The central focus of this study is the systematic evaluation of EEG channel reduction.

The following configurations are compared:

- 32 channels → Full-channel baseline
- 16 channels → 50% channel reduction
- 8 channels → 75% channel reduction
- 4 channels → 87.5% channel reduction

The 32-channel configuration is treated as the baseline for each model and emotion.

The analysis investigates whether reducing the number of EEG channels necessarily causes a proportional decline in performance.

```python
all_results["Reduction_Percentage"] = (
    (32 - all_results["Channels"]) / 32 * 100
)

display(
    all_results[
        [
            "Model",
            "Feature",
            "Emotion",
            "Channels",
            "Reduction_Percentage",
            "Accuracy"
        ]
    ]
)
```

```python
baseline = (
    all_results[
        all_results["Channels"] == 32
    ][
        [
            "Model",
            "Feature",
            "Emotion",
            "Accuracy",
            "Balanced_Accuracy",
            "F1",
            "ROC_AUC"
        ]
    ]
    .rename(
        columns={
            "Accuracy": "Accuracy_32",
            "Balanced_Accuracy": "Balanced_Accuracy_32",
            "F1": "F1_32",
            "ROC_AUC": "ROC_AUC_32"
        }
    )
)

channel_analysis = all_results.merge(
    baseline,
    on=["Model", "Feature", "Emotion"],
    how="left"
)

channel_analysis["Accuracy_Change"] = (
    channel_analysis["Accuracy"] -
    channel_analysis["Accuracy_32"]
)

channel_analysis["F1_Change"] = (
    channel_analysis["F1"] -
    channel_analysis["F1_32"]
)

display(channel_analysis)
```

# Key Channel Reduction Findings

The experimental results demonstrate that EEG channel reduction does not have a uniform effect across all emotional dimensions or model pipelines.

### Valence

The DGCNN achieves its highest accuracy using 32 channels (55.36%).

However, the Mean-Variance Random Forest improves from 45.71% with 32 channels to 56.79% with only 16 channels.

The Band-Power Random Forest achieves its highest Valence accuracy of 53.57% using 8 channels.

This demonstrates that the optimal channel count depends on the feature representation and model.

### Arousal

The DGCNN achieves its best result with 32 channels (52.86%).

The Band-Power Random Forest achieves 53.21% using only 4 channels, compared with 53.57% using 32 channels.

Thus, the Band-Power pipeline retains nearly the same classification accuracy despite an 87.5% reduction in EEG channels.

### Dominance

Dominance is more sensitive to channel reduction for the DGCNN.

DGCNN accuracy decreases from 51.43% with 32 channels to 32.86% with 8 channels.

In contrast, both Random Forest pipelines achieve their highest Dominance accuracy using 32 channels, reaching 59.29%.

This indicates that Dominance may require more spatial EEG information under the evaluated experimental conditions.

### Liking

The Band-Power Random Forest achieves its highest Liking accuracy of 63.93% using only 8 channels.

This is slightly higher than its 32-channel performance of 63.57%.

Therefore, a 75% reduction in channel count does not result in performance degradation for this configuration.

The DGCNN also achieves its best Liking performance using 16 channels (57.86%), rather than 32 channels (57.14%).

These results provide strong evidence that the full 32-channel configuration is not universally optimal.

# Main Findings from the Proposed Models

The experimental results reveal several important observations.

## 1. No Single Model Is Optimal for Every Emotion

The best model and channel configuration vary across Valence, Arousal, Dominance, and Liking.

This indicates that EEG emotion recognition is strongly dependent on the emotional dimension being predicted.

## 2. More Channels Do Not Always Mean Better Performance

Several reduced-channel configurations perform similarly to or better than their 32-channel counterparts.

For example:

- Mean-Variance RF for Valence:
  45.71% → 56.79% when reducing from 32 to 16 channels.

- Band-Power RF for Liking:
  63.57% → 63.93% when reducing from 32 to 8 channels.

- Band-Power RF for Arousal:
  53.57% with 32 channels and 53.21% with only 4 channels.

These results demonstrate that channel reduction can sometimes preserve or improve classification performance.

## 3. Feature Representation Influences Channel Requirements

The optimal channel configuration differs between Mean-Variance and Band-Power representations.

This indicates that the information carried by EEG channels is not independent of the feature extraction strategy.

## 4. Model Complexity Does Not Guarantee Better Performance

The DGCNN does not consistently outperform the simpler Random Forest pipelines.

For example, the Random Forest models achieve stronger results for several Dominance and Liking configurations.

This suggests that a more complex deep-learning architecture does not automatically guarantee better performance under the evaluated experimental conditions.

## 5. Dominance Appears More Sensitive to Channel Reduction

The DGCNN performance for Dominance decreases substantially when reducing channels.

This contrasts with Liking, where strong performance is maintained even after substantial channel reduction.

Therefore, different emotional dimensions may have different spatial information requirements.

## 6. Channel Reduction Can Be Emotion-Specific

The experiments demonstrate that the optimal channel count is not universal.

Instead, it depends on:

- Emotional dimension
- Feature representation
- Model architecture

# Practical Significance

The channel-reduction results have potential practical implications for EEG-based emotion recognition.

A conventional EEG system with 32 channels may be:

- More expensive
- More difficult to configure
- Less portable
- More time-consuming to place on a user

If comparable performance can be achieved using fewer channels, reduced-electrode systems may become more practical.

The experiments demonstrate that:

- 16 channels represent a 50% reduction.
- 8 channels represent a 75% reduction.
- 4 channels represent an 87.5% reduction.

Importantly, the results indicate that the feasibility of channel reduction depends on the emotional dimension and model pipeline.

Therefore, the study does not propose that all EEG systems should universally use four or eight channels.

Instead, it demonstrates the need for systematic evaluation to determine the appropriate channel configuration for a particular emotion recognition task.

# Final Conclusion

## Overall Conclusion

This study presents a systematic investigation of EEG channel reduction for emotion recognition using the DEAP-derived dataset.

Three different pipelines were evaluated under a common experimental framework:

1. DGCNN using graph-based EEG representations
2. Random Forest using Mean-Variance statistical features
3. Random Forest using Band-Power frequency-domain features

Each pipeline was evaluated using 32, 16, 8, and 4 EEG channels across four emotional dimensions: Valence, Arousal, Dominance, and Liking.

The results demonstrate that the relationship between EEG channel count and emotion-recognition performance is not linear or universal.

Increasing the number of EEG channels does not always result in improved classification accuracy. In several cases, reduced-channel configurations achieved performance comparable to or better than the full 32-channel configuration.

For example, the Random Forest model using Mean-Variance features improved Valence accuracy from 45.71% with 32 channels to 56.79% with 16 channels, corresponding to a 50% reduction in channel count.

Similarly, the Random Forest Band-Power model achieved its highest Liking accuracy of 63.93% using only 8 channels, slightly exceeding its 32-channel accuracy of 63.57%. This represents a 75% reduction in EEG channels without a reduction in classification accuracy.

For Arousal, the Band-Power Random Forest achieved 53.21% accuracy with only four channels, compared with 53.57% using 32 channels. Although the difference is small, this result demonstrates that substantial channel reduction may be possible for specific emotional dimensions and feature representations.

In contrast, Dominance showed greater sensitivity to channel reduction, particularly for the DGCNN model. Its accuracy decreased from 51.43% with 32 channels to 32.86% with 8 channels. This indicates that the optimal number of EEG channels depends strongly on the emotional dimension.

The comparison between DGCNN and Random Forest models also demonstrates that increased model complexity does not necessarily guarantee superior performance. The simpler Random Forest pipelines achieved competitive or superior performance for several emotional dimensions and channel configurations.

Therefore, the results highlight an important relationship between:

    Model Architecture
            +
    Feature Representation
            +
    EEG Channel Count
            +
    Emotional Dimension

The optimal channel configuration cannot be determined independently of these factors.

## Limitations of the Proposed Models

The proposed EEG emotion recognition models have several limitations that should be considered when interpreting the results:

- **Limited Dataset Size:** The experiments are conducted on the DEAP dataset, which contains data from a relatively limited number of subjects. This may restrict the generalizability of the models to larger and more diverse populations.

- **Subject Variability:** EEG signals exhibit significant inter-subject variability. Although the DGCNN model uses a subject-independent evaluation setup, performance may still vary across individuals due to differences in brain activity patterns and emotional responses.

- **Channel Reduction Trade-off:** Reducing the number of EEG channels decreases the hardware and computational requirements but may also result in the loss of useful spatial information. Therefore, aggressive channel reduction may negatively affect classification performance.

- **Dataset and Experimental Constraints:** The models are evaluated on a single dataset and under a specific experimental protocol. Their performance may differ when applied to EEG data collected using different devices, electrode configurations, or experimental environments.

- **Model-Specific Limitations:** The DGCNN model has higher computational complexity and training requirements compared with the Random Forest models. In contrast, the Random Forest models depend on manually extracted statistical or frequency-domain features, which may not capture all complex spatial and temporal relationships present in raw EEG signals.

- **Class Distribution:** The distribution of emotion classes can influence model performance, particularly for binary emotion classification. Therefore, accuracy alone may not fully represent model performance, and metrics such as balanced accuracy, precision, recall, F1-score, and ROC-AUC should also be considered.

- **Limited Direct Comparison with Existing Literature:** The existing studies use different preprocessing pipelines, channel configurations, feature extraction methods, and evaluation protocols. Consequently, comparisons with published results should be interpreted as reference benchmarks rather than strictly equivalent head-to-head comparisons.

- **Generalization to Real-World Applications:** Further validation on independent datasets and real-world EEG recordings is required to determine whether the proposed models can maintain their performance outside the DEAP experimental setting.
